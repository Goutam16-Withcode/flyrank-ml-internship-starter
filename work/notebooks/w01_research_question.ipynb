This notebook frames our core research question and explores initial empirical evidence from the FlyRank dataset.


## 1. My lane (or freestyle) and why

**Selected Lane**: **Content Refresh Prioritization (SEO Decay Prevention)**

**Rationale**: Organic search traffic decays over time due to content staleness, search engine algorithm updates, and emerging competitor pages. Editorial teams have limited bandwidth and cannot update every page simultaneously. Prioritizing content refresh using empirical ranking signals maximizes organic search recovery per editor-hour spent.


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

raw_data_path = ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df_raw = pd.read_csv(raw_data_path)
print(f"Loaded dataset: {len(df_raw):,} rows, {len(df_raw.columns)} columns")
```

## 2. The question: decision, action, cost of a wrong call

- **Decision**: Which published web pages are currently exhibiting organic search performance decay and require immediate editorial re-optimization?
- **Action**: Place high-risk decaying pages into a prioritized weekly content update queue for technical and editorial refresh.
- **Cost of a wrong call**:
  - *False Positive (updating a healthy page)*: Wastes valuable editorial resources ($150-$300 per review) without yielding traffic recovery.
  - *False Negative (missing a decaying page)*: Leads to compounding loss of search engine rankings, organic traffic, and revenue over time.


```python
declining_count = (df_raw['trend_direction'] == 'down').sum()
total_count = len(df_raw)
pct_declining = (declining_count / total_count) * 100

print(f"Total Pages Analyzed: {total_count:,}")
print(f"Pages with Downward Trend: {declining_count:,} ({pct_declining:.1f}%)")
print(f"Estimated Monthly At-Risk Traffic: {df_raw[df_raw['trend_direction']=='down']['sessions_90d'].sum() / 3:,.0f} sessions/month")
```

## 3. Quick look at the data (2-3 real numbers)

Key empirical baseline observations from the starter dataset:
1. **Total Dataset Volume**: 30,000 pseudonymized pages across diverse clients.
2. **Decay Baseline Rate**: **34.5%** of pages exhibit a downward search trend (`trend_direction == 'down'`).
3. **Traffic Impact**: At-risk declining pages represent over **1.4 Million sessions per month** that can be salvaged with timely content updates.
