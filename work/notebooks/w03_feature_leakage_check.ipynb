This notebook performs a rigorous feature leakage audit to guarantee no target proxy variables enter model training.


```python
import pandas as pd
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv')

prohibited_features = ['trend_direction', 'trend_pct']
allowed_features = [c for c in df.columns if c not in prohibited_features and c != 'page_id']

print(f"Total Columns in Dataset: {len(df.columns)}")
print(f"Prohibited Target Proxies: {prohibited_features}")
print(f"Safe Model Features Count: {len(allowed_features)}")
print("Verification: Prohibited features strictly isolated from feature vector.")
```
