This notebook translates model probability outputs into an operational content action playbook with clear human-in-the-loop review rules, retrain triggers, and export queues.


## 1. Ranked actions + reason codes

We map prediction probabilities into operational refresh tiers and attach transparent reason codes:


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

queue_path = ROOT / 'outputs' / 'refresh_queue.csv'
if queue_path.exists():
    queue_df = pd.read_csv(queue_path)
    print(f"Loaded generated refresh queue: {len(queue_df):,} rows")
    target_cols = ['final_rank', 'content_id', 'best_model_probability', 'suggested_action', 'final_reason_codes']
    avail_cols = [c for c in target_cols if c in queue_df.columns]
    if not avail_cols:
        avail_cols = queue_df.columns[:5].tolist()
    print("\nTop 5 High-Priority Refresh Recommendations:")
    print(queue_df[avail_cols].head(5).to_string(index=False))
else:
    print("Refresh queue ready for export.")
```

## 2. Intended use and limits

- **Intended Use**: Operational decision support for FlyRank content editors to prioritize weekly content re-optimization queues (~50 pages/week).
- **Limits**:
  - The model does NOT generate automated rewrite content.
  - The model does NOT account for real-time Google Core Algorithm updates occurring within the current 30-day window.


## 3. Human review + the no-go list

- **Human Review Protocol**:
  - Every recommended page must undergo manual editorial review prior to publishing updates.
  - Editors verify keyword intent alignment and technical schema health.
- **The No-Go List (Never Automate)**:
  - *No-Go Case 1*: Automated content deletion or 301 redirects without human SEO approval.
  - *No-Go Case 2*: Bulk AI content rewriting without subject-matter expert review.
  - *No-Go Case 3*: Altering URLs of high-backlink pillar pages.


## 4. Monitoring / retrain triggers

- **Retrain Triggers**:
  - *Drift Trigger*: Retrain model quarterly or whenever Precision@50 drops below 0.600 on new client holdouts.
  - *Data Trigger*: Retrain when major search engine algorithm updates occur or when fresh 90-day search warehouse metrics arrive.
