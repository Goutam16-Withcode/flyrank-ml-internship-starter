This notebook translates model probability outputs into actionable business recommendations and ranked queue exports.


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv')

# Load pipeline outputs
queue_path = ROOT / 'outputs' / 'refresh_queue.csv'
if queue_path.exists():
    queue_df = pd.read_csv(queue_path)
    print(f"Loaded generated refresh queue: {len(queue_df):,} rows")
    print("\nTop 5 High-Priority Refresh Recommendations:")
    print(queue_df[['rank', 'page_id', 'model_score', 'recommended_action', 'primary_reason']].head(5).to_string(index=False))
else:
    print("Refresh queue ready for export.")
```
