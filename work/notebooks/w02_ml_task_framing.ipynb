This notebook defines the machine learning task framing, prediction unit, target label, and primary evaluation metrics.


## 1. Machine Learning Formulation

- **Task Type**: Binary Supervised Classification & Learning-to-Rank.
- **Unit of Prediction**: Individual URL / Content ID (`content_id`).
- **Target Label**: `is_declining_label` defined as:
  $$\text{is\_declining\_label} = \begin{cases} 1 & \text{if } \text{trend\_direction} = \text{'down'} \\ 0 & \text{otherwise} \end{cases}$$


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv')
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)

print("Target Label Distribution:")
print(df['is_declining_label'].value_counts(normalize=True).rename({1: 'Declining (1)', 0: 'Stable/Growing (0)'}))
```

## 2. Baseline vs ML Framing & Metrics

- **Primary Evaluation Metric**: **Precision@50** (Precision among the top 50 predicted pages).
- **Secondary Evaluation Metric**: Precision@100 and ROC-AUC.
- **Why Precision@50?**: Editorial teams operate under capacity constraints (e.g. reviewing ~50 pages per week). Maximizing the proportion of truly declining pages within the top 50 recommendations directly optimizes business ROI.


```python
random_p50 = df['is_declining_label'].mean()
print(f"Random Guessing Precision@50: {random_p50:.3f}")
print("Target Benchmark for ML Model: >0.700 Precision@50 (~3x lift over baseline)")
```
