This notebook evaluates bivariate signal strength and correlation between feature candidates and search decay.


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv')
df['target'] = (df['trend_direction'] == 'down').astype(int)

numeric_cols = ['days_since_last_update', 'content_age_days', 'ctr', 'avg_position', 'impressions_90d', 'engagement_rate']
correlations = df[numeric_cols].apply(lambda x: x.corr(df['target'])).sort_values(ascending=False)

print("Top Numeric Feature Correlations with Target (is_declining_label):")
for col, corr in correlations.items():
    print(f"  {col:<25}: {corr:+.4f}")
```
