This notebook implements a transparent hand-written rule baseline score to establish the benchmark for ML models.


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('.').resolve()
while not (ROOT / 'data').exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

df = pd.read_csv(ROOT / 'data' / 'raw' / 'content_refresh_anonymized.csv')
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)

# Rule Baseline Scoring Function
def compute_baseline_score(row):
    score = 0.0
    if row['days_since_last_update'] > 180:
        score += 0.35
    if row['impressions_90d'] > 100:
        score += 0.25
    if row['ctr'] < 0.02:
        score += 0.20
    if row['avg_position'] > 15:
        score += 0.20
    return score

df['baseline_score'] = df.apply(compute_baseline_score, axis=1)

# Evaluate Baseline Precision@50
top50 = df.sort_values('baseline_score', ascending=False).head(50)
baseline_p50 = top50['is_declining_label'].mean()

print(f"Rule Baseline Precision@50: {baseline_p50:.3f}")
print("Hand-written heuristic established. Next step: Train ML models to achieve >3x lift.")
```
